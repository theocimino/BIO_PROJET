### **Question 2 - Adimensionnalisation**

On considère le système initial:

\[
\begin{cases}
\dot{P} = r(H)P - \delta P^2 \\
\dot{H} = \alpha P H - m H
\end{cases}
\]

On cherche des changements de variables pour obtenir un système de la forme :

$$
\begin{cases}
\dot{P} = r(H)P - \delta P^2 \\
\dot{H} = P H - H
\end{cases}
$$

On pose les variables sans dimension :

$$
\begin{aligned}
&x = k_1P, \qquad y = k_2H, \qquad \tau = k_3 t .
\end{aligned}
$$

**Calcul de $\dfrac{dx}{d\tau}$**

- Développement :

$$
\begin{aligned}
&\frac{dx}{d\tau}
= \frac{d(k_1P)}{d(k_3t)}
= \frac{k_1}{k_3}\frac{dP}{dt} \\
&= \frac{k_1}{k_3}(r(H)P-\delta P^2) \\
&= \frac{k_1}{k_3}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\frac{x}{k_1}
- \delta\Bigl(\frac{x}{k_1}\Bigr)^2\Bigr] \\
&= \frac{1}{k_3}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \Bigl(\frac{\delta}{k_1}\Bigr) x^2\Bigr].
\end{aligned}
$$

- Donc :

$$
\begin{aligned}
&\frac{dx}{d\tau} = \frac{1}{k_3}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \Bigl(\frac{\delta}{k_1}\Bigr) x^2\Bigr].
\end{aligned}
$$

**Calcul de $\dfrac{dy}{d\tau}$**

- Développement :

$$
\begin{aligned}
&\frac{dy}{d\tau}
= \frac{d(k_2H)}{d(k_3t)}
= \frac{k_2}{k_3}\frac{dH}{dt} \\
&= \frac{k_2}{k_3}(\alpha PH-mH) \\
&= \frac{k_2}{k_3}\Bigl[\alpha \frac{x}{k_1}\frac{y}{k_2}
- m\frac{y}{k_2}\Bigr] \\
&= \frac{1}{k_3}\Bigl[\frac{\alpha}{k_1}xy-my\Bigr].
\end{aligned}
$$

On pose $k_3 = m$ et $k_1 = \alpha/m$, alors :

$$
\begin{aligned}
&\frac{dy}{d\tau}
= \frac{1}{m}\Bigl[\frac{\alpha}{\alpha/m}xy-my\Bigr]
= xy-y.
\end{aligned}
$$

On réinjecte $k_3=m$ et $k_1=\alpha/m$ dans l’expression de $\dfrac{dx}{d\tau}$ :

$$
\begin{aligned}
&\frac{dx}{d\tau}
= \frac{1}{m}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \frac{\delta}{\alpha/m}x^2\Bigr] \\
&= \frac{1}{m}r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \frac{\delta}{\alpha}x^2.
\end{aligned}
$$
On peut conserver $\delta = \frac{\delta}{\alpha}$ sans conséquence car $\alpha$ positif et $\delta$ non fixé.
En posant $r_1(z)=\frac{1}{m}r(z)$ et $k_2=1$, on obtient :

$$
\begin{aligned}
&\frac{dx}{d\tau} = r_1(y)x - \delta x^2, \\
&\frac{dy}{d\tau} = xy - y.
\end{aligned}
$$

Finalement, en renommant les variables, le système adimensionnalisé est de la forme :

$$
\begin{cases}
\dot{P} = r_1(H)P - \delta P^2, \\
\dot{H} = PH - H,
\end{cases}
$$

avec $r_1(H) = \frac{1}{m}r(H)$.

Dans la suite, on pourra noter $r_1(H)$ comme $r(H)$ sans ambiguïté, étant donné que le scaling $\frac{1}{m}$ est absorbé par r0 qui n'est jamais fixé.

```julia
using Plots
#Définition des variables,fonctions pour plus tard

#CONSTANTES
r0 = 1
β = 1/6
γ = 1

function r(H)
   return r0 * ( 1 + H / β ) * ( 1 - H / γ )
end

function model_adim(P,H, δ=1)
    return [r(H) * P - δ * P^2 ; P * H - H]
end
```

### **Question 4 - Jacobienne**

on a $J(P,H) = 
\begin{pmatrix}
r(H) - 2\delta P & P r'(H) \\
H & P - 1
\end{pmatrix}$   

Pour chaque cas, on évalue la jacobienne en chacun des équilibres en espérant obtenir des informations sur la stabilité de ceux-ci grâce au théorème de Lyapunov indirect.

**Rappels sur Lyapunov indirect :**    
On évalue la jacobienne liée au système pour chaque équilibre $X^* = (P^*,H^*)$, et :    
- si toutes les valeurs propres de $Jac(X^*)$ sont à partie réelle négative alors $X^*$ est un équilibre asymptotiquement stable (ASS) ie :     
        si $\forall i,\ \operatorname{Re}(\lambda_i) < 0\ \text{avec}\ \lambda_i \in \sigma(Jac(X^*)) \implies X^*\ \text{asymptotiquement stable}$
- s'il existe une valeur propre de $Jac(X^*)$ à partie réelle strictement positive alors $X^*$ est instable ie :    
        si $\exists i, \ \operatorname{Re}(\lambda_i) > 0\ \text{avec}\ \lambda_i \in \sigma(Jac(X^*)) \implies X^*\ \text{instable}$
- si toutes les parties réelles des valeurs sont négatives à l'exception de certaines qui sont égales à 0, on ne peut pas conclure ie :   
 si $\exists i, \ \operatorname{Re}(\lambda_i)=0\ \text{et}\ \forall j\neq i,\ \operatorname{Re}(\lambda_j)<0 \quad \implies \quad \text{stabilité de } X^* \text{ indéterminée}$



- ###  Equilibres en présence agnostique aux cas : 
On note 2 équilibres, en (P,H) respectivement :    
    - en $(0,0)$   
    - en $(\frac{r(0)}{\delta},0)$

Peu importe le cas étudié (1,3 ou 4), ces équilbres sont présents dans le système, déterminons leur stabilité.



- Pour $(0,0)$ :   
$J(0,0) = 
\begin{pmatrix}
r(0) & 0 \\
0 & -1
\end{pmatrix}$    
Les valeurs propres sont $r(0)$ et $-1$ or $r(0) \in \mathbb{R_+^*}$ donc $\operatorname{Re}(r(0)) > 0$, ainsi $(0,0)$ est instable.    

- Pour  $(\frac{r(0)}{\delta},0)$ :   
$J(\frac{r(0)}{\delta},0) = 
\begin{pmatrix}
-r(0) & \frac{1}{\delta} r(0) r'(0) \\
0 & \frac{r(0)}{\delta}-1
\end{pmatrix}$    
Les valeurs propres d'une matrice triangulaire supérieure sont également les termes sur la diagonale.   
On pose :   
$\lambda_1 = -r(0)$ et $\lambda_2 = \frac{r(0)}{\delta}-1$
Donc si $\operatorname{Re}(\lambda_1)$ et $\operatorname{Re}(\lambda_2)$ sont toutes 2 inférieures à 0, on a stabilité de  $(\frac{r(0)}{\delta},0)$.

C'est à dire dans le cas 1 et le cas 3, $(\frac{r(0)}{\delta},0)$ est asymptotiquement stable.

### Tout d'abord, on a $\frac{(\gamma + \beta)^2}{4\gamma\beta}r(0) > r0$.

En effet, on a $(\frac{(\gamma + \beta)^2}{4\gamma\beta} > 1) \Leftrightarrow ( (\gamma + \beta)^2 > 4\gamma\beta )$

En développant on obtient $(\gamma^2 + 2\gamma\beta + \beta^2 > 4 \gamma\beta) \Leftrightarrow (\gamma^2 - 2\gamma\beta + \beta^2 > 0)$

En remarquant que $\gamma^2 - 2\gamma\beta + \beta^2 = (\gamma - \beta)^2$, on a donc $$(\frac{(\gamma + \beta)^2}{4\gamma\beta}r(0) > r0) \Leftrightarrow ((\gamma - \beta)^2 > 0),$$ 
ce qui est toujours vrai.


Dans la suite on va donc étudier les équilibres dans les cas suivants : $\delta \in [0, r(0)]$, $\delta \in [r(0), \frac{(\gamma + \beta)^2}{4\gamma\beta}r(0)]$, $\delta \in [\frac{(\gamma + \beta)^2}{4\gamma\beta}r(0), +\infty[$ qui correspondent respectivement aux cas 4, cas 3, et cas 1.

### On considère d'abord le segment $[0, r(0)]$ :



À l'équilibre non trivial, on a 
$$
r(H^*) = \delta
$$

En utilisant l'expression générale de $r(H)$ donnée par :
$$
r(H) = r(0) \left(1 + \frac{H}{\beta}\right) \left(1 - \frac{H}{\gamma}\right)
$$

L'équation devient :
$$
r(0) \left( 1 - \frac{H^*}{\gamma} + \frac{H^*}{\beta} - \frac{(H^*)^2}{\beta\gamma} \right) = \delta
$$

On obtient une équation du second degré en $H^*$:
$$
-\frac{1}{\gamma\beta}{H^*}^2 + {H^*}(\frac{1}{\beta} - \frac{1}{\gamma}) + 1 - \frac{\delta}{r(0)} = 0
$$
$$
\Leftrightarrow
$$
$$
{H^*}^2 - (\gamma - \beta)H^* - \beta\gamma \left( 1 - \frac{\delta}{r(0)} \right) = 0
$$

Calculons le discriminant $\Delta$ de ce polynôme :
$$
\Delta = (\gamma - \beta)^2 + 4\beta\gamma \left( 1 - \frac{\delta}{r(0)} \right)
$$

Par hypothèse, nous avons par hypothèse $\delta \leq r(0)$, ce qui implique que le terme $\left( 1 - \frac{\delta}{r(0)} \right)$ est positif ou nul.

Par conséquent, on a toujours $\Delta > 0$ : l'équation admet deux racines réelles distinctes.

On a 
$$
H_{\pm}^* = \frac{(\gamma - \beta) \pm \sqrt{\Delta}}{2}
$$

(Pour rappel, on a $\gamma > \beta$ quel que soit le cas) 

Très clairement, on a $H_+ > H_-$ et les analyses précédentes nous montrent que $H_-$ est une racine négative et $H_+$ une racine positive.

Les densités de population doivent être positives ou nulles ($H \geq 0$), donc l'expression de cet équilibre que l'on nomme $H_2^*$ en fonction de $\delta$ est la suivante :
$$
H_2^*(\delta) = \frac{(\gamma - \beta) + \sqrt{(\gamma - \beta)^2 + 4\beta\gamma \left( 1 - \frac{\delta}{r(0)} \right)}}{2}
$$

### Analyse qualitative

*   **Limite quand ($\delta \to 0^+$) :** par continuité on a égalité avec
    $$
    H_2^*(0) = \frac{(\gamma - \beta) + \sqrt{(\gamma - \beta)^2 + 4\beta\gamma}}{2}
    $$
    En remarquant que $(\gamma - \beta)^2 + 4\beta\gamma = (\gamma + \beta)^2$, on obtient :
    $$
    H_2^*(0) = \frac{(\gamma - \beta) + (\gamma + \beta)}{2} = \gamma
    $$

*   **Limite quand ($\delta \to r(0)^-$) :** par continuité on a égalité avec
    $$
    H_2^*(r(0)) = \frac{(\gamma - \beta) + \sqrt{(\gamma - \beta)^2}}{2}
    $$
    
    $$
    H_2^*(r(0)) = \frac{(\gamma - \beta) + (\gamma - \beta)}{2} = \gamma - \beta < \gamma
    $$
    
On a donc $H_2^*(\delta)$ qui vaut $\gamma$ en $0$ et qui décroit de manière concave, par composition de $ \delta \mapsto (\gamma - \beta)^2 + 4\beta\gamma ( 1 - \frac{\delta}{r(0)} )$ (linéaire décroissante) et $x \mapsto \sqrt{x}$ (concave croissante), jusqu'a atteindre $\gamma - \beta$ en $r(0)$




### On considère maintenant le segment $[r(0), \frac{(\gamma + \beta)^2}{4\gamma\beta}r(0)] = [r(0), r(H_m)]$



On se ramène à la même équation que précedemment, mais cette fois on a $\delta > r(0)$, on sait cependant que il y a deux équilibres positifs donc on peut supposer $\Delta$ positif

$$
H_{\pm}^* = \frac{(\gamma - \beta) \pm \sqrt{\Delta}}{2}
$$

avec
$$
\Delta = (\gamma - \beta)^2 + 4\beta\gamma \left( 1 - \frac{\delta}{r(0)} \right)
$$

On va se référer à $H_-^*$ en tant que $H_1^*$ et $H_+^*$ comme $H_2^*$ on sait que ces deux se rejoignent en $H_m = \frac{\gamma - \beta}{2}$  

Analysons ces équilibres :

- $H_1^*(\delta) = \frac{(\gamma - \beta) - \sqrt{(\gamma - \beta)^2 + 4\beta\gamma \left( 1 - \frac{\delta}{r(0)} \right)}}{2}$  
On a d'une part $H_1^*(r(0)) = \frac{(\gamma - \beta) - (\gamma - \beta)}{2} = 0 $ \
D'autre part, on cherche pour quel $\delta$ on a $H_1^*(\delta)$ qui vaut $H_m$, on résout alors :
$$
\frac{(\gamma - \beta) - \sqrt{(\gamma - \beta)^2 + 4\beta\gamma \left( 1 - \frac{\delta}{r(0)} \right)}}{2} = H_m = \frac{\gamma - \beta}{2}
$$
Ce qui revient bien à annuler $\Delta$ \
En résolvant l'équation, on obtient $\delta = r(0)(1 + \frac{(\gamma - \beta)^2}{4\gamma\beta}) = r(0)\frac{(\gamma + \beta)^2}{4\gamma\beta} = r(H_m)$
Ce qui est cohérant, car cela revient au cas où $\frac{r(H_m)}{\delta} = 1$ \
On en conclut que $H_1^*(r(0)) = 0$ et par des arguments similaires au cas précédent (continuité, composition...), $H_1^*$ est croissante de manière convexe jusqu'a atteindre $\frac{\gamma - \beta}{2}$ en $r(H_m)$




- $H_2^*(\delta) = \frac{(\gamma - \beta) + \sqrt{(\gamma - \beta)^2 + 4\beta\gamma \left( 1 - \frac{\delta}{r(0)} \right)}}{2}$ \
On a $H_2^*(r(0)) = \frac{(\gamma - \beta) + (\gamma - \beta)}{2} = \gamma - \beta > H_m = \frac{\gamma - \beta}{2}$

$H_2^*$ vaut $\gamma - \beta$ en $r(0)$ et est décroissante de manière concave jusqu'a atteindre $H_m = \frac{\gamma - \beta}{2}$

### On considère maintenant $[\frac{(\gamma + \beta)^2}{4\gamma\beta}r(0), +\infty[ = [r(H_m), +\infty[$



Dans ce cas il n'y a pas d'équilibres non triviaux pour H, donc rien à tracer au-delà de $\delta = \frac{(\gamma + \beta)^2}{4\gamma\beta}r(0) = r(H_m)$

```julia
# Calcul du maximum de r(H) pour les limites du tracé
Hm = (γ - β) / 2
r_max = r(Hm)

# Fonctions définissant les branches d'équilibres H*
function H1_star(δ)
    return ((γ - β) - sqrt((γ - β)^2 + 4 * β * γ * (1 - δ / r0))) / 2
end

function H2_star(δ)
    return ((γ - β) + sqrt((γ - β)^2 + 4 * β * γ * (1 - δ / r0))) / 2
end

# Création des vecteurs de données pour delta
delta_cas4 = range(0.01, r0, length=150)
delta_cas3 = range(r0, r_max, length=150)

# On choisit les valeurs numériques à marquer, et on leur associe un texte
valeurs_x = [0.0, r0, r_max]
labels_x = ["0", "r(0)", "r_max"]

valeurs_y = [0.0, Hm, γ - β, γ]
labels_y = ["0", "H_m", "γ - β", "γ"]

# Tracé du diagramme complet avec xticks et yticks
p_bifurcation = plot(
    xlabel="Paramètreδ", 
    ylabel="Équilibres H*",
    title="Diagramme de bifurcation de H* en fonction de δ",
    grid=true,
    framestyle=:box,
    legend=:bottomleft,
    xlims=(0, r_max + 0.1),
    ylims=(-0.1, γ + 0.1),
    xticks=(valeurs_x, labels_x), # On remplace les graduations par les lettres
    yticks=(valeurs_y, labels_y)  # Pareil pour l'axe Y
)

# -- Cas 4 : Branche H2*
plot!(p_bifurcation, delta_cas4, H2_star.(delta_cas4), 
      label="H₂* (Cas 4 : δ ≤ r₀)", color=:darkgreen, linewidth=2.5)

# -- Cas 3 : Branche H2*
plot!(p_bifurcation, delta_cas3, H2_star.(delta_cas3), 
      label="H₂* (Cas 3 : r₀ < δ ≤ r_max)", color=:limegreen, linewidth=2.5)

# -- Cas 3 : Branche H1* 
plot!(p_bifurcation, delta_cas3, H1_star.(delta_cas3), 
      label="H₁* instable (Cas 3)", color=:red, linestyle=:dash, linewidth=2.5)


# Ajout des points d'intérêt
scatter!(p_bifurcation, [r0], [0.0], color=:magenta, markersize=5, label="Bifurcation Transcritique")
scatter!(p_bifurcation, [r_max], [Hm], color=:black, markersize=5, label="Bifurcation Nœud-Col")
# Point de départ sur l'axe des ordonnées

display(p_bifurcation)
```
