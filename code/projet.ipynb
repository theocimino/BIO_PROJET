### **Question 2 - Adimensionnalisation**

On considère le système initial:

\[
\begin{cases}
\dot{P} = r(H)P - \delta P^2 \\
\dot{H} = \alpha P H - m H
\end{cases}
\]

On cherche des changements de variables pour obtenir un système de la forme :

$$
\begin{cases}
\dot{P} = r(H)P - \delta P^2 \\
\dot{H} = P H - H
\end{cases}
$$

On pose les variables sans dimension :

$$
\begin{aligned}
&x = k_1P, \qquad y = k_2H, \qquad \tau = k_3 t .
\end{aligned}
$$

**Calcul de $\dfrac{dx}{d\tau}$**

- Développement :

$$
\begin{aligned}
&\frac{dx}{d\tau}
= \frac{d(k_1P)}{d(k_3t)}
= \frac{k_1}{k_3}\frac{dP}{dt} \\
&= \frac{k_1}{k_3}(r(H)P-\delta P^2) \\
&= \frac{k_1}{k_3}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\frac{x}{k_1}
- \delta\Bigl(\frac{x}{k_1}\Bigr)^2\Bigr] \\
&= \frac{1}{k_3}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \Bigl(\frac{\delta}{k_1}\Bigr) x^2\Bigr].
\end{aligned}
$$

- Donc :

$$
\begin{aligned}
&\frac{dx}{d\tau} = \frac{1}{k_3}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \Bigl(\frac{\delta}{k_1}\Bigr) x^2\Bigr].
\end{aligned}
$$

**Calcul de $\dfrac{dy}{d\tau}$**

- Développement :

$$
\begin{aligned}
&\frac{dy}{d\tau}
= \frac{d(k_2H)}{d(k_3t)}
= \frac{k_2}{k_3}\frac{dH}{dt} \\
&= \frac{k_2}{k_3}(\alpha PH-mH) \\
&= \frac{k_2}{k_3}\Bigl[\alpha \frac{x}{k_1}\frac{y}{k_2}
- m\frac{y}{k_2}\Bigr] \\
&= \frac{1}{k_3}\Bigl[\frac{\alpha}{k_1}xy-my\Bigr].
\end{aligned}
$$

On pose $k_3 = m$ et $k_1 = \alpha/m$, alors :

$$
\begin{aligned}
&\frac{dy}{d\tau}
= \frac{1}{m}\Bigl[\frac{\alpha}{\alpha/m}xy-my\Bigr]
= xy-y.
\end{aligned}
$$

On réinjecte $k_3=m$ et $k_1=\alpha/m$ dans l’expression de $\dfrac{dx}{d\tau}$ :

$$
\begin{aligned}
&\frac{dx}{d\tau}
= \frac{1}{m}\Bigl[r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \frac{\delta}{\alpha/m}x^2\Bigr] \\
&= \frac{1}{m}r\!\Bigl(\frac{y}{k_2}\Bigr)\,x
- \frac{\delta}{\alpha}x^2.
\end{aligned}
$$
On peut conserver $\delta = \frac{\delta}{\alpha}$ sans conséquence car $\alpha$ positif et $\delta$ non fixé.
En posant $r_1(z)=\frac{1}{m}r(z)$ et $k_2=1$, on obtient :

$$
\begin{aligned}
&\frac{dx}{d\tau} = r_1(y)x - \delta x^2, \\
&\frac{dy}{d\tau} = xy - y.
\end{aligned}
$$

Finalement, en renommant les variables, le système adimensionnalisé est de la forme :

$$
\begin{cases}
\dot{P} = r_1(H)P - \delta P^2, \\
\dot{H} = PH - H,
\end{cases}
$$

avec $r_1(H) = \frac{1}{m}r(H)$.


```julia
#Définition des variables,fonctions pour plus tard

#CONSTANTES
r0 = 1
β = 1/6
γ = 1

function r(H)
   return r0 * ( 1 + H / β ) * ( 1 - H / γ )
end

function model_adim(P,H, δ=1)
    return [r(H) * P - δ * P^2 ; P * H - H]
end
```

### **Question 3 - Isoclines nulles**

$$
r(H)P - \delta P^2 = 0 \quad \Rightarrow \quad P \bigl( r(H) - \delta P \bigr) = 0

```julia
H_range = 0:0.01:2

####################################
# 1) Fenêtre pour l'isocline P' = 0
####################################
iso_P1 = [(h, r(h)) for h in H_range if r(h) ≥ 0]   # P = r(H)
iso_P2 = [(h, 0.0)   for h in H_range]              # P = 0

pP = plot(
    xlabel = "H", ylabel = "P",
    title = "Isoclines P' = 0",
    xlims = (0, 2), ylims = (0, 2.5),
    legend = false,
    grid = true
)

# P = r(H)
plot!(pP,
    first.(iso_P1), last.(iso_P1),
    color = :blue, linewidth = 3,
)

# P = 0
plot!(pP,
    first.(iso_P2), last.(iso_P2),
    color = :blue, linewidth = 3,
)

display(pP)
```

### **Question 4 - Jacobienne**

on a $J(P,H) = 
\begin{pmatrix}
r_1(H) - 2\delta P & P r_1'(H) \\
H & P - 1
\end{pmatrix}$   

Pour chaque cas, on évalue la jacobienne en chacun des équilibres en espérant obtenir des informations sur la stabilité de ceux-ci grâce au théorème de Lyapunov indirect.

**Rappels sur Lyapunov indirect :**    
On évalue la jacobienne liée au système pour chaque équilibre $X^* = (P^*,H^*)$, et :    
- si toutes les valeurs propres de $Jac(X^*)$ sont à partie réelle négative alors $X^*$ est un équilibre asymptotiquement stable (ASS) ie :     
        si $\forall i,\ \Re(\lambda_i) < 0\ \text{avec}\ \lambda_i \in \sigma(Jac(X^*)) \implies X^*\ \text{asymptotiquement stable}$
- s'il existe une valeur propre de $Jac(X^*)$ à partie réelle strictement positive alors $X^*$ est instable ie :    
        si $\exists i, \ \Re(\lambda_i) > 0\ \text{avec}\ \lambda_i \in \sigma(Jac(X^*)) \implies X^*\ \text{instable}$
- si toutes les parties réelles des valeurs sont négatives à l'exception de certaines qui sont égales à 0, on ne peut pas conclure ie :   
 si $\exists i, \ \Re(\lambda_i)=0\ \text{et}\ \forall j\neq i,\ \Re(\lambda_j)<0 \quad \implies \quad \text{stabilité de } X^* \text{ indéterminée}$



- ###  Equilibres en présence agnostique aux cas : 
On note 2 équilibres, en (P,H) respectivement :    
    - en $(0,0)$   
    - en $(\frac{r_1(0)}{\delta},0)$

Peu importe le cas étudié (1,3 ou 4), ces équilbres sont présents dans le système, déterminons leur stabilité.



- Pour $(0,0)$ :   
$J(0,0) = 
\begin{pmatrix}
r_1(0) & 0 \\
0 & -1
\end{pmatrix}$    
Les valeurs propres sont $r_1(0)$ et $-1$ or $r_1(0) \in \mathbb{R_+^*}$ donc $\Re(r_1(0)) > 0$, ainsi $(0,0)$ est instable.    

- Pour  $(\frac{r_1(0)}{\delta},0)$ :   
$J(\frac{r_1(0)}{\delta},0) = 
\begin{pmatrix}
-r_1(0) & \frac{1}{\delta} r_1(0) r_1'(0) \\
0 & \frac{r_1(0)}{\delta}-1
\end{pmatrix}$    
Les valeurs propres d'une matrice triangulaire supérieure sont également les termes sur la diagonale.   
On pose :   
$\lambda_1 = -r_1(0)$ et $\lambda_2 = \frac{r_1(0)}{\delta}-1$
Donc si $\Re(\lambda_1)$ et $\Re(\lambda_2)$ sont toutes 2 inférieures à 0, on a stabilité de  $(\frac{r_1(0)}{\delta},0)$.   
Déterminons les différents cas qui découlent de cette assertion.   
Par définition $r_1(0)$ est positif donc $\Re(\lambda_1)<0$.   
Et $\Re(\lambda_2)<0 \iff \lambda_2 <0\iff \frac{r_1(0)}{\delta}-1 < 0 \iff \delta > r_1(0) $   
Cette condition est vérifiée dans les cas 1 et 3.   
C'est à dire dans le cas 1 et le cas 3, $(\frac{r_1(0)}{\delta},0)$ est asymptotiquement stable.   
En revanche, dans le cas 4, $(\frac{r_1(0)}{\delta},0)$ est instable.

- cas 3 : présence de 2 équilibres positifs
On évalue la jacobienne en $H_1^*$ et en $H_2^*$
En ces équilibres, P vaut nécessairement 1 (cf graphs précédents) donc $P(H_1^*) = P(H_2^*) = 1$

### Rappel sur un critère de convergence
En plaçant sur un plan le determinant de la jacobienne évaluée aux points d'équilibres et la trace évaluée en ces mêmes points, on peut être en capacité de définir les types de stabilité et d'instabilité de ces équilibres.

![Critère de stabilité](../input/crit_stab.png)


Or,   
$det[J(P(H_1^*),H_1^*)] = - H_1^* r_1'(H_1^*)$   
Or en $H_1^*$ équilibre positif atteint avant le max de $r_1$, on a $r_1$ croissante car parabole concave donc :    
$r_1'(H_1^*) > 0 ⟹ det[J(P(H_1^*),H_1^*)] < 0$    
Donc ici, $(tr[J(P(H_1^*),H_1^*)], det[J(P(H_1^*),H_1^*)])$ est dans la zone de point selle (instable).

- Pour $(P(H_2^*),H_2^*)$ :   
$J(P(H_2^*),H_2^*) = 
\begin{pmatrix}
r_1(H_2^*) - 2\delta P(H_2^*) & P(H_2^*) r_1'(H_2^*) \\
H_2^* & P(H_2^*) - 1
\end{pmatrix}   = 
\begin{pmatrix}
P(H_2^*)\delta - 2\delta &  r_1'(H_2^*) \\
H_2^* & 0
\end{pmatrix} =
\begin{pmatrix}
-\delta &  r_1'(H_2^*) \\
H_2^* & 0
\end{pmatrix}
$     
Donc $tr(J(P(H_2^*),H_2^*)) = -\delta < 0$   
et $det(J(P(H_2^*),H_2^*)) = -H_2^* r_1'(H-2^*)$   
Or $h_2^*$ équilibre positif et $r_1'(H-2^*)$ dérivée d'une parabole concave post-maximum nécessairement négative donc :    
$det(J(P(H_2^*),H_2^*)) > 0$    
Au final $(tr[J(P(H_2^*),H_2^*)], det[J(P(H_2^*),H_2^*)])$ se trouve dans le quart nord-est du graph, stable !
